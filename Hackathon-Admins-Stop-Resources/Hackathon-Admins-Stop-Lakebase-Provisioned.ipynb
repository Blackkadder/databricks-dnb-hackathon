```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()
```

```python
lakebase_p_list = [lb for lb in w.database.list_database_instances()]
for lbp in lakebase_p_list:
    print(lbp.name)
```

```python
from databricks.sdk.service.database import DatabaseInstanceState, DatabaseInstance

for lb_instance in lakebase_p_list:
    if lb_instance.state == DatabaseInstanceState.AVAILABLE:
        w.database.update_database_instance(
            name=lb_instance.name,
            database_instance=DatabaseInstance(name=lb_instance.name, stopped=True),
            update_mask="*",
        )
        print(f"Stopped database instance: {lb_instance.name}")
    elif lb_instance.state == DatabaseInstanceState.STOPPED:
        print(f"{lb_instance.name} already STOPPED")
```
