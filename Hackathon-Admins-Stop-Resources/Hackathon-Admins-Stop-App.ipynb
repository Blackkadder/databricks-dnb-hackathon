```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()
```

```python
apps_list = [dbapp for dbapp in w.apps.list()]
for app in apps_list:
    print(app.name)
```

```python
from databricks.sdk.service.apps import ComputeState

for dbapp in apps_list:
    if dbapp.compute_status.state == ComputeState.ACTIVE:
        w.apps.stop(name=dbapp.name)
    elif dbapp.compute_status.state == ComputeState.STOPPED:
        print(f"{dbapp.name} already STOPPED")
```
